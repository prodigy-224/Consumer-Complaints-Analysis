# **Consumer Complaints Analysis**

# Business Problem

Financial institutions receive complaints from consumers about different financial products and
services. These complaints may involve account management, credit cards, mortgages, credit
reporting, payment services and other financial activities. This analysis focuses on identifying the perfomance of these complaints over time, the products with the most complaints, also get to see whether the organizations offered timely responses and more.

```python
# Standardizing the column names
df.columns = df.columns.str.replace(" ", "_")

# Adding Year and Month columns from Date_submitted
df['Year'] = df['Date_submitted'].dt.year
df['Month'] = df['Date_submitted'].dt.strftime('%B')
```

1. **Total Complaints**

This refers to the total number of complaints that were raised

```python
df.shape[0]
```

There are a total of 62,516 complaints

2. **Annual Complaint Volume**

This refers to the total number of complaints by year

```python
df.groupby("Year").size()
```

2017 had the least number of complaints while 2022 and 2021 had the highest number of complaints

3. **Monthly Complaint Volume**

Refers to the total number of complaints per month

```python
df.groupby('Month').size().sort_values(ascending=False)
```

July and August have the most number of complaints while Feb has the least

4. **Complaints by Financial Product**

Refers to the number of complaints by each product

```python
df.groupby("Product").size().sort_values(ascending=False)
```

checking or savings account has the most number of complaints followed by credit card or prepaid card while student loans has the least number of complaints

7. **Percentage contribution of each product to total complaints.**

```python
(df['Product'].value_counts(normalize=True) * 100).round(2)
```

8. **Company response category distribution.**

```python
df['Company_response_to_consumer'].value_counts()
```

9. **Number of complaints marked In progress.**

```python
In_progress = df["Company_response_to_consumer"] == "In progress"
print(In_progress.sum())
```

10. **Percentage of complaints marked as receiving a timely response.**

```python
(df['Timely_response?'].value_counts(normalize=True, dropna=False) * 100).round(2)
```

93% of the complaints get a timely response

11. **Average and median complaint-forwarding interval.**

```python
forwarding_days = (df['Date_received'] - df['Date_submitted']).dt.days
print(f"Average forwarding days: {forwarding_days.mean():.2f} days")
print(f"Median forwarding days: {forwarding_days.median()} days")
```
